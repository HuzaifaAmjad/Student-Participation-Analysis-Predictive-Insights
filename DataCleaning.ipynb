```python
import pandas as pd
#load raw data
df_raw=pd.read_csv("RIT+Opportunity+Wise+Data+-+Sheet1.csv")
#display whole dataset
print("="*100)
df_raw
```

```python
#making copy of raw data for preserving original data
df=df_raw.copy()
df.info()
```

```python
#count missing values in each column
print("Missing values per column")
print("="*100)
missing_values=df.isnull().sum()
print(missing_values)
#total missing values
print(f"\nTotal missing values: {missing_values.sum()}")
```

```python
print("Statistical Summary:")
print("="*100)
df.describe()
```

```python
import pandas as pd
import numpy as np
#Remove duplicate rows
df.drop_duplicates(inplace=True)
#handle missing values
numerical_cols=df.select_dtypes(include=['int64','float64']).columns
for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())
#categorical columns, fill with mode
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    if not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])
# Correct data format
date_columns=[
    'Opportunity End Date',
    'Date of Birth',
    'Entry created at',
    'Apply Date',
    'Opportunity Start Date'
]
for col in date_columns:
    if col in df.columns:
        # Convert to datetime
        df[col] = pd.to_datetime(df[col], errors='coerce')
#fill invalid/missing dates with the most frequent date
        if not df[col].mode().empty:
            df[col] = df[col].fillna(df[col].mode()[0])

# ----------------------------
# Clean and Standardize Text Data
# ----------------------------

# Keep only the first word in the First Name column
df['First Name'] = df['First Name'].str.split().str[0]

# Remove leading and trailing spaces from all text columns
text_columns = df.select_dtypes(include=['object']).columns

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

# Standardize capitalization
df['Gender'] = df['Gender'].str.title()
df['Country'] = df['Country'].str.title()
df['Opportunity Category'] = df['Opportunity Category'].str.title()
df['Institution Name'] = df['Institution Name'].str.title()

# Fix Gender inconsistencies
df['Gender'] = df['Gender'].replace({
    "Don'T Want To Specify": "Don't Want To Specify"
})

# Fix Institution Name inconsistencies
df['Institution Name'] = df['Institution Name'].replace({
    "Saint Louis": "Saint Louis University",
    "St Louis University": "Saint Louis University",
    "Saintlouis University": "Saint Louis University"
})
df["Institution Name"] = df["Institution Name"].replace({
    "St. Louis University": "Saint Louis University"
})

# Merge all variations containing "Saint Louis"
df.loc[
    df['Institution Name'].str.contains("Saint Louis", case=False, na=False),
    'Institution Name'
] = "Saint Louis University"

# Remove Outliers in numeric columns
for col in numerical_cols:
    Q1=df[col].quantile(0.25)
    Q3=df[col].quantile(0.75)
    IQR=Q3-Q1
    lower=Q1-1.5*IQR
    upper=Q3+1.5*IQR
    df = df[(df[col] >= lower) & (df[col] <= upper)]
# Validate data
print("\nMissing Values:")
print(df.isnull().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
print("\nData Types:")
print(df.dtypes)
# Save cleaned dataset
print("\nFinal Shape:", df.shape)
print(df.head())
df.to_csv('Student_Participation_Cleaned.csv',index=False)
```
